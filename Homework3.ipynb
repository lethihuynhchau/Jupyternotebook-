```python
import pandas as pd
import numpy as np

# 1. Load the CSV dataset
df = pd.read_csv("perceptron_toy_dataset.csv")

# 2. Insert the bias column
df.insert(0, "x0", 1.0)

# 3. Display the dataset
print("--- Dataset with Added Bias Term (x0 = 1.0) ---")
print(df)
```

```python

# Prepare the data for training
X = df[['x0', 'feature_x1', 'feature_x2']].to_numpy().reshape(-1, 3, 1)
y = df['label_y'].to_numpy()

# Initialize weights and learning parameters
w = np.array([[1, -0.5, -1]])
eta = 0.1
epochs = 5

# Train the perceptron
for epoch in range(1, epochs + 1):
    print(f"\n--- Epoch {epoch} (Online Perceptron Updates) ---")
    misclassified_count = 0

    for i in range(len(X)):
        x_i = X[i]
        y_i = y[i]
        z = np.dot(w, x_i)[0, 0]

        if y_i * z <= 0:
            misclassified_count += 1
            w = w + eta * y_i * x_i.T
            print(
                f"Sample {i+1:2d} (y={y_i:2d}, z={z:5.2f}): "
                f"MISCLASSIFIED -> Updated w: {w[0]}"
            )
        else:
            print(
                f"Sample {i+1:2d} (y={y_i:2d}, z={z:5.2f}): "
                "Correct -> Weights unchanged"
            )

    accuracy_rate = ((len(X) - misclassified_count) / len(X)) * 100

    print(f"\nEpoch {epoch} complete with {misclassified_count}/{len(X)} misclassification(s).")
    print(f"Accuracy Rate: {accuracy_rate:.2f}%")
    print(f"Final Weights after Epoch {epoch}: {w[0]}")

    if misclassified_count == 0:
        print("\nConvergence achieved! Perfect separation (100% accuracy) reached.")
        break
```

```python

# Define new samples
new_samples = np.array([
    [2.0, 2.0],
    [-2.0, -2.0],
    [0.0, 0.0]
])

# Add bias term x0 = 1
new_samples_with_bias = np.c_[np.ones(len(new_samples)), new_samples]

# Predict labels using the trained weights
predictions = np.sign(new_samples_with_bias @ w.T).flatten()

print("New Samples and Predicted Labels:")
for sample, prediction in zip(new_samples, predictions):
    print(f"Sample {sample} -> Predicted label: {int(prediction)}")
```
