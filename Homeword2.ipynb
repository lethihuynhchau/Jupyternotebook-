```python

class Person:
    def __init__(self, name, age, occupation):
        self.name = name
        self.age = age
        self.birthyear = 2025 - age
        self.occupation = occupation

    def greeting(self):
        print('Hello. I am ' + self.name + '. I am ' + str(self.age)
              + '. I was born in ' + str(self.birthyear)
              + '. I work as a ' + self.occupation + '.')

    def get_age_in_days(self):
        return self.age * 365


# Task 1a
person_1 = Person('Lucky', 21, 'Software Engineer')
person_1.greeting()

# Task 1b
person_0 = Person('Lucky', 21, 'Student')
days = person_0.get_age_in_days()
print(person_0.name + ' is approximately ' + str(days) + ' days old.')
```

```python

class raceCar:
    def __init__(self, _v):
        self.pos = 0
        self.v = _v

    def run(self, _t):
        runTime = _t
        self.pos = self.pos + runTime * self.v

    def showCurrentPos(self):
        print('Pos = ', self.pos)
        
    def reverse(self, t):
        self.pos = self.pos - t * self.v

rc = raceCar(30)  # 30 km/h

rc.showCurrentPos()
rc.run(1)
rc.showCurrentPos()

rc.run(0.5)
rc.showCurrentPos()

rc.reverse(0.5)
rc.showCurrentPos()

        
```

```python

import numpy as np

cm = np.array([
    [12, 3],
    [2, 15]
])

print(cm)

def compute_metrics(cm):
    TN = cm[0, 0]
    FP = cm[0, 1]
    FN = cm[1, 0]
    TP = cm[1, 1]

    return TP, TN, FP, FN

TP, TN, FP, FN = compute_metrics(cm)

print("TP =", TP)
print("TN =", TN)
print("FP =", FP)
print("FN =", FN)

def compute_rates(TP, TN, FP, FN):
    TPR = TP / (TP + FN)
    TNR = TN / (TN + FP)
    return TPR, TNR

TPR, TNR = compute_rates(TP, TN, FP, FN)

print("TPR =", round(TPR, 4))
print("TNR =", round(TNR, 4))

accuracy = (TP + TN) / np.sum(cm)

print("Accuracy =", round(accuracy, 4))
print("TPR =", round(TPR * 100, 2), "%")
print("TNR =", round(TNR * 100, 2), "%")
```

```python
import pandas as pd
from sklearn.linear_model import Perceptron
from sklearn.metrics import confusion_matrix, accuracy_score

df = pd.read_csv('Bank_dataset.csv')

print(df.head())
print(df.columns.tolist())
print(df.shape)
```

```python
X = df[['debt_income_ratio', 'job_position', 'working_period']]
y = df['loan_status']

model = Perceptron()
model.fit(X, y)

y_pred = model.predict(X)

car = accuracy_score(y, y_pred)
cm = confusion_matrix(y, y_pred, labels=[-1, 1])

print("CAR =", round(car, 4))
print("CAR (%) =", round(car * 100, 2), "%")
print("Confusion Matrix (labels = [-1, 1]):")
print(cm)
```

```python
new_customer = pd.DataFrame({
    'debt_income_ratio': [0.45],
    'job_position': [0],
    'working_period': [0]
})

prediction = model.predict(new_customer)

print("Predicted loan_status =", prediction[0])
```

```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import Perceptron
from sklearn.metrics import accuracy_score, confusion_matrix

df = pd.read_csv('Bank_dataset.csv')

print(df.head())
print(df.shape)
print(df.columns.tolist())
```

```python
X = df[['debt_income_ratio', 'job_position', 'working_period']]
y = df['loan_status']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=1
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
model = Perceptron()
model.fit(X_train, y_train)

y_train_pred = model.predict(X_train)

train_car = accuracy_score(y_train, y_train_pred)
train_cm = confusion_matrix(y_train, y_train_pred, labels=[-1, 1])

print("Training CAR =", round(train_car, 4))
print("Training CAR (%) =", round(train_car * 100, 2), "%")
print("Training Confusion Matrix (labels = [-1, 1]):")
print(train_cm)
```

```python
y_test_pred = model.predict(X_test)

test_car = accuracy_score(y_test, y_test_pred)
test_cm = confusion_matrix(y_test, y_test_pred, labels=[-1, 1])

print("Testing CAR =", round(test_car, 4))
print("Testing CAR (%) =", round(test_car * 100, 2), "%")
print("Testing Confusion Matrix (labels = [-1, 1]):")
print(test_cm)
```

```python
car_values = [train_car * 100, test_car * 100]
labels = ['Training CAR', 'Testing CAR']

plt.figure(figsize=(7, 5))
plt.bar(labels, car_values)
plt.ylabel('Classification Accuracy Rate (%)')
plt.title('Training CAR vs Testing CAR')
plt.ylim(0, 100)

for i, value in enumerate(car_values):
    plt.text(i, value + 1, f'{value:.2f}%', ha='center')

plt.show()
```
