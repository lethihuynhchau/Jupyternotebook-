```python
# ==========================================
# STEP 2: FEATURE EXTRACTION
# ==========================================

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# 1. Resize image
def resize_nearest(img, new_h, new_w):
    old_h, old_w = img.shape[:2]

    row_idx = np.linspace(0, old_h - 1, new_h).astype(int)
    col_idx = np.linspace(0, old_w - 1, new_w).astype(int)

    grid_y, grid_x = np.meshgrid(
        row_idx, col_idx, indexing='ij'
    )

    return img[grid_y, grid_x]


# 2. Compute Roundness
def compute_Roundness(mask):
    area = np.sum(mask)

    padded = np.pad(
        mask, pad_width=1,
        mode='constant',
        constant_values=0
    )

    perimeter_mask = (
        mask &
        (
            (~padded[:-2, 1:-1]) |
            (~padded[2:, 1:-1]) |
            (~padded[1:-1, :-2]) |
            (~padded[1:-1, 2:])
        )
    )

    perimeter = np.sum(perimeter_mask)

    if perimeter > 0:
        roundness = 4 * np.pi * area / (perimeter ** 2)
    else:
        roundness = 0

    return roundness


# 3. Compute Elongation
def compute_Elongation(mask):
    ys, xs = np.where(mask)

    if len(xs) < 2:
        return 1.0

    coords = np.column_stack((xs, ys))

    cov = np.cov(coords, rowvar=False)

    eigvals = np.linalg.eigvalsh(cov)

    if eigvals[0] > 0:
        elongation = np.sqrt(eigvals[1]) / np.sqrt(eigvals[0])
    else:
        elongation = np.inf

    return elongation


# 4. Feature Extraction
def FeatureExtraction(img_raw):

    if img_raw.dtype != np.uint8:
        img_raw = (img_raw * 255).astype(np.uint8)

    if img_raw.shape[2] == 4:
        img_raw = img_raw[:, :, :3]

    img = resize_nearest(img_raw, 200, 200)

    # Convert to grayscale
    gray = np.dot(
        img[..., :3],
        [0.2989, 0.5870, 0.1140]
    ).astype(np.uint8)

    # Otsu threshold
    hist, bin_edges = np.histogram(
        gray.ravel(),
        bins=256,
        range=(0, 256)
    )

    total = gray.size
    current_max = 0
    threshold = 0

    sum_total = np.dot(np.arange(256), hist)

    sum_background = 0.0
    weight_background = 0.0

    for i in range(256):

        weight_background += hist[i]

        if weight_background == 0:
            continue

        weight_foreground = total - weight_background

        if weight_foreground == 0:
            break

        sum_background += i * hist[i]

        mean_background = sum_background / weight_background

        mean_foreground = (
            sum_total - sum_background
        ) / weight_foreground

        between_var = (
            weight_background * weight_foreground *
            (mean_background - mean_foreground) ** 2
        )

        if between_var > current_max:
            current_max = between_var
            threshold = i

    # Apply threshold
    mask = gray < threshold

    # Calculate shape features
    roundness = compute_Roundness(mask)
    elongation = compute_Elongation(mask)

    # Calculate average RGB
    object_pixels = img[mask]

    if object_pixels.size > 0:
        avg_r, avg_g, avg_b = object_pixels.mean(axis=0)
    else:
        avg_r, avg_g, avg_b = -1, -1, -1

    featureVector = np.array([
        roundness,
        elongation,
        avg_r,
        avg_g,
        avg_b
    ])

    return featureVector

print("All feature extraction functions are ready!")
```

```python
papaya_features = FeatureExtraction(papaya_img)
tomato_features = FeatureExtraction(tomato_img)

feature_names = [
    "Roundness",
    "Elongation",
    "Avg_R",
    "Avg_G",
    "Avg_B"
]

print("PAPAYA FEATURES")
for name, value in zip(feature_names, papaya_features):
    print(f"{name}: {value:.4f}")

print("\nTOMATO FEATURES")
for name, value in zip(feature_names, tomato_features):
    print(f"{name}: {value:.4f}")
```

```python
# ==========================================
# STEP 2: CREATE DATASET
# ==========================================

dataset = []

class_folders = ["tomato", "papaya"]
n_images = 10

for class_name in class_folders:

    for i in range(n_images):

        img_path = os.path.join(
            dataset_path,
            class_name,
            f"image{i}.bmp"
        )

        try:
            img_raw = plt.imread(img_path)

            feature_vec = FeatureExtraction(img_raw)

            # Assign class label
            if class_name == "tomato":
                classcode = 0
            else:
                classcode = 1

            row = feature_vec.tolist() + [classcode]

            dataset.append(row)

            print(f"Processed {img_path}")

        except Exception as e:
            print(f"Failed {img_path}: {e}")


# Create DataFrame
columns = [
    "Roundness",
    "Elongation",
    "Avg_R",
    "Avg_G",
    "Avg_B",
    "Label"
]

df = pd.DataFrame(dataset, columns=columns)

# Export CSV
csv_path = os.path.join(
    dataset_path,
    "tomato_vs_papaya_dataset.csv"
)

df.to_csv(csv_path, index=False)

print("\nDataset created successfully!")
print("Dataset shape:", df.shape)
print("Saved at:", csv_path)
```

```python
# ==========================================
# STEP 3: TRAIN LINEAR PERCEPTRON
# ==========================================

from sklearn.linear_model import Perceptron
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, confusion_matrix

# Load dataset
df = pd.read_csv(csv_path)

# Separate features and labels
X_train = df[
    ['Roundness', 'Elongation', 'Avg_R', 'Avg_G', 'Avg_B']
].values

y_train = df['Label'].values

# Feature scaling
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)

# Initialize and train Perceptron
model = Perceptron(max_iter=1000, random_state=42)
model.fit(X_train_scaled, y_train)

print("Perceptron model trained successfully!")
```

```python
# Predict training data
y_pred_train = model.predict(X_train_scaled)

# Accuracy
train_acc = accuracy_score(y_train, y_pred_train)

# Confusion Matrix
cm = confusion_matrix(y_train, y_pred_train)

tn, fp, fn, tp = cm.ravel()

# Calculate TPR and TNR
tpr = tp / (tp + fn) if (tp + fn) > 0 else 0.0
tnr = tn / (tn + fp) if (tn + fp) > 0 else 0.0

print("=== Training Evaluation Metrics ===")
print(f"Training Accuracy Rate: {train_acc * 100:.2f}%")
print(f"Confusion Matrix:\n{cm}")
print(f"True Positive Rate (TPR / Sensitivity): {tpr:.4f}")
print(f"True Negative Rate (TNR / Specificity): {tnr:.4f}")
```

```python
import os
import numpy as np
import pandas as pd
from PIL import Image

classes = {
    "apple": 0,
    "banana": 1,
    "orange": 2
}

data = []

for class_name, label in classes.items():
    folder = os.path.join(dataset_path, class_name)

    image_files = sorted([
        f for f in os.listdir(folder)
        if f.lower().endswith(".bmp")
    ])[:5]

    for image_file in image_files:
        image_path = os.path.join(folder, image_file)

        img = np.array(Image.open(image_path).convert("RGB"))

        features = FeatureExtraction(img)

        data.append([
            features[0],
            features[1],
            features[2],
            features[3],
            features[4],
            label
        ])

df_abo = pd.DataFrame(data, columns=[
    "Roundness",
    "Elongation",
    "Avg_R",
    "Avg_G",
    "Avg_B",
    "Label"
])

csv_path = os.path.join(
    dataset_path,
    "a_b_o_classification_dataset.csv"
)

df_abo.to_csv(csv_path, index=False)

print("Dataset created successfully!")
print("Dataset shape:", df_abo.shape)
print("\nNumber of samples per class:")
print(df_abo["Label"].value_counts().sort_index())

display(df_abo)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.multiclass import OneVsRestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix

# Select input features and labels
X_train = df_abo[
    ["Roundness", "Elongation", "Avg_R", "Avg_G", "Avg_B"]
].values

y_train = df_abo["Label"].values

# Normalize features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)

# Logistic Regression with One-vs-Rest
base_model = LogisticRegression(
    solver="liblinear",
    random_state=42
)

model = OneVsRestClassifier(base_model)

# Train model
model.fit(X_train_scaled, y_train)

# Predict training data
y_pred_train = model.predict(X_train_scaled)

# Calculate accuracy
train_acc = accuracy_score(y_train, y_pred_train)

# Confusion Matrix
cm = confusion_matrix(
    y_train,
    y_pred_train,
    labels=[0, 1, 2]
)

print("Training Accuracy Rate: {:.2f}%".format(train_acc * 100))

print("\nConfusion Matrix:")
print(cm)

print("\nClass labels:")
print("0 = Apple")
print("1 = Banana")
print("2 = Orange")
```

```python
import os
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# -----------------------------------------------------------
def resize_nearest(img, new_h, new_w):
    old_h, old_w = img.shape[:2]
    row_idx = (np.linspace(0, old_h - 1, new_h)).astype(int)
    col_idx = (np.linspace(0, old_w - 1, new_w)).astype(int)
    grid_y, grid_x = np.meshgrid(row_idx, col_idx, indexing='ij')
    return img[grid_y, grid_x]

# -----------------------------------------------------------
def compute_Roundness(mask):
    area = np.sum(mask)
    padded = np.pad(mask, pad_width=1, mode='constant', constant_values=0)
    perimeter_mask = (
        mask &
        (
            (~padded[:-2, 1:-1]) |
            (~padded[2:, 1:-1])  |
            (~padded[1:-1, :-2]) |
            (~padded[1:-1, 2:])
        )
    )
    perimeter = np.sum(perimeter_mask)
    if perimeter > 0:
        roundness = 4 * np.pi * area / (perimeter ** 2)
    else:
        roundness = 0
    return roundness

# -----------------------------------------------------------
def compute_Elongation(mask):
    ys, xs = np.where(mask)
    if len(xs) == 0:
        return 1.0
    coords = np.column_stack((xs, ys))
    cov = np.cov(coords, rowvar=False)
    eigvals = np.linalg.eigvalsh(cov)
    if eigvals[0] > 0:
        elongation = np.sqrt(eigvals[1]) / np.sqrt(eigvals[0])
    else:
        elongation = np.inf
    return elongation

# -----------------------------------------------------------
def FeatureExtraction(img_raw):
    if img_raw.dtype != np.uint8:
        img_raw = (img_raw * 255).astype(np.uint8)
    if img_raw.shape[2] == 4:
        img_raw = img_raw[:, :, :3]
    img = resize_nearest(img_raw, 200, 200)

    gray = np.dot(img[..., :3], [0.2989, 0.5870, 0.1140]).astype(np.uint8)

    # Otsu's threshold
    hist, bin_edges = np.histogram(gray.ravel(), bins=256, range=(0, 256))
    total = gray.size
    current_max, threshold = 0, 0
    sum_total = np.dot(np.arange(256), hist)
    sum_background, weight_background = 0.0, 0.0
    for i in range(256):
        weight_background += hist[i]
        if weight_background == 0:
            continue
        weight_foreground = total - weight_background
        if weight_foreground == 0:
            break
        sum_background += i * hist[i]
        mean_background = sum_background / weight_background
        mean_foreground = (sum_total - sum_background) / weight_foreground
        between_var = (weight_background * weight_foreground *
                       (mean_background - mean_foreground) ** 2)
        if between_var > current_max:
            current_max = between_var
            threshold = i

    mask = gray < threshold

    roundness = compute_Roundness(mask)
    elongation = compute_Elongation(mask)

    object_pixels = img[mask]
    if object_pixels.size > 0:
        avg_r, avg_g, avg_b = object_pixels.mean(axis=0)
    else:
        avg_r, avg_g, avg_b = -1, -1, -1

    return np.array([roundness, elongation, avg_r, avg_g, avg_b])

# -----------------------------------------------------------
def CreateDataset(ImgeNum=5):
    class_folders = ['apple', 'banana', 'orange']
    label_mapping = {'apple': 0, 'banana': 1, 'orange': 2}
    dataset = []

    for class_name in class_folders:
        folder_path = os.path.join(homework_path, class_name)
        for i in range(ImgeNum):
            img_path = os.path.join(folder_path, f'image{i}.bmp')
            try:
                img_raw = plt.imread(img_path)
                feature_vec = FeatureExtraction(img_raw)
                classcode = label_mapping[class_name]
                row = feature_vec.tolist() + [classcode]
                dataset.append(row)
                print(f"Processed {img_path} -> Label: {classcode}")
            except Exception as e:
                print(f"Failed {img_path}: {e}")

    columns = ['Roundness', 'Elongation', 'Avg_R', 'Avg_G', 'Avg_B', 'Label']
    df = pd.DataFrame(dataset, columns=columns)

    csv_path = os.path.join(homework_path, 'a_b_o_classification_dataset.csv')
    df.to_csv(csv_path, index=False)
    print(f"\nSuccessfully exported dataset to {csv_path}")
    return df

df = CreateDataset(ImgeNum=5)
print()
print(df.round(3))
```

```python
import os
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.multiclass import OneVsRestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, confusion_matrix

# 1. Đọc dataset
csv_path = os.path.join(homework_path, 'a_b_o_classification_dataset.csv')
df = pd.read_csv(csv_path)

features = ['Roundness', 'Elongation', 'Avg_R', 'Avg_G', 'Avg_B']
X_train = df[features].values
y_train = df['Label'].values

# Kiểm tra dữ liệu hợp lệ (không NaN, không inf)
print("Số mẫu:", X_train.shape[0], "| Số đặc trưng:", X_train.shape[1])
print("Có NaN/inf không?", not np.isfinite(X_train).all())

# 2. Chuẩn hóa đặc trưng
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)

# 3. Huấn luyện Logistic Regression (One-vs-Rest)
model = OneVsRestClassifier(
    LogisticRegression(solver='liblinear', random_state=42)
)
model.fit(X_train_scaled, y_train)

# 4. Đánh giá trên tập huấn luyện
y_pred_train = model.predict(X_train_scaled)
train_acc = accuracy_score(y_train, y_pred_train)
cm = confusion_matrix(y_train, y_pred_train, labels=[0, 1, 2])

print("\n=== Training Evaluation Metrics (Logistic Regression OvR) ===")
print(f"Training Accuracy Rate: {train_acc * 100:.2f}%")
print("Confusion Matrix:\n", cm)
```

```python
import os
import random
import numpy as np
import matplotlib.pyplot as plt

label_map = {0: 'Apple', 1: 'Banana', 2: 'Orange'}
classes_info = [
    {'name': 'Apple',  'label': 0, 'dir': os.path.join(homework_path, 'apple')},
    {'name': 'Banana', 'label': 1, 'dir': os.path.join(homework_path, 'banana')},
    {'name': 'Orange', 'label': 2, 'dir': os.path.join(homework_path, 'orange')},
]
valid_extensions = ('.bmp',)   # chỉ lấy .bmp, bỏ qua thư mục [Originals]

sample_results = []
for c in classes_info:
    files = [f for f in os.listdir(c['dir']) if f.lower().endswith(valid_extensions)]
    if not files:
        raise FileNotFoundError(f"Không có ảnh trong {c['dir']}")

    random_file = random.choice(files)
    img_raw = plt.imread(os.path.join(c['dir'], random_file))

    # Trích đặc trưng (dùng lại hàm ở Bước 2)
    features = FeatureExtraction(img_raw).reshape(1, -1)

    # Ảnh để hiển thị (cùng kích thước 200x200 như khi trích đặc trưng)
    img_show = img_raw
    if img_show.dtype != np.uint8:
        img_show = (img_show * 255).astype(np.uint8)
    img_show = resize_nearest(img_show[:, :, :3], 200, 200)

    # Chuẩn hóa bằng scaler đã fit rồi dự đoán
    scaled = scaler.transform(features)
    pred_label = int(model.predict(scaled)[0])
    pred_probs = model.predict_proba(scaled)[0]

    sample_results.append({
        'img': img_show,
        'filename': random_file,
        'true_name': c['name'],
        'true_label': c['label'],
        'pred_label': pred_label,
        'pred_name': label_map[pred_label],
        'confidence': pred_probs[pred_label] * 100,
    })

# Vẽ kết quả
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
for ax, res in zip(axes, sample_results):
    ax.imshow(res['img'])
    ax.set_title(
        f"File: {res['filename']}\n"
        f"True Class: {res['true_name']} ({res['true_label']})\n"
        f"Predicted: {res['pred_name']} ({res['pred_label']})\n"
        f"Confidence: {res['confidence']:.1f}%"
    )
    ax.axis('off')
plt.tight_layout()
plt.show()
```

```python
import os
import random
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, random_split, Subset
from torchvision import transforms
from torchvision.datasets import ImageFolder
from sklearn.metrics import classification_report

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

dataset_path = os.path.join(homework_path, 'Car_vs_Truck_Large_Dataset')

# -------------------------------------------------------------
# 1. Biến đổi ảnh (train có augmentation, val thì không)
# -------------------------------------------------------------
train_transforms = transforms.Compose([
    transforms.Resize((128, 128)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(degrees=15),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

val_transforms = transforms.Compose([
    transforms.Resize((128, 128)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

# -------------------------------------------------------------
# 2. Chia train/val = 80/20
# -------------------------------------------------------------
full_dataset = ImageFolder(root=dataset_path)
class_names = full_dataset.classes
print("Các lớp:", class_names, "| Tổng số ảnh:", len(full_dataset))

train_size = int(0.8 * len(full_dataset))
val_size = len(full_dataset) - train_size
train_subset, val_subset = random_split(
    range(len(full_dataset)), [train_size, val_size],
    generator=torch.Generator().manual_seed(42)
)

train_dataset = Subset(ImageFolder(root=dataset_path, transform=train_transforms), train_subset.indices)
val_dataset = Subset(ImageFolder(root=dataset_path, transform=val_transforms), val_subset.indices)

train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False)
print("Train:", len(train_dataset), "ảnh | Val:", len(val_dataset), "ảnh")

# -------------------------------------------------------------
# 3. Mô hình CNN có Dropout
# -------------------------------------------------------------
class RegularizedCNN(nn.Module):
    def __init__(self):
        super(RegularizedCNN, self).__init__()

        self.conv1 = nn.Conv2d(3, 16, 3, padding=1)
        self.bn1 = nn.BatchNorm2d(16)
        self.relu1 = nn.ReLU()
        self.pool1 = nn.MaxPool2d(2, 2)

        self.conv2 = nn.Conv2d(16, 32, 3, padding=1)
        self.bn2 = nn.BatchNorm2d(32)
        self.relu2 = nn.ReLU()
        self.pool2 = nn.MaxPool2d(2, 2)

        self.conv3 = nn.Conv2d(32, 64, 3, padding=1)
        self.bn3 = nn.BatchNorm2d(64)
        self.relu3 = nn.ReLU()
        self.pool3 = nn.MaxPool2d(2, 2)

        self.flatten = nn.Flatten()
        self.fc1 = nn.Linear(64 * 16 * 16, 128)
        self.relu4 = nn.ReLU()
        self.dropout = nn.Dropout(p=0.5)
        self.fc2 = nn.Linear(128, 2)

    def forward(self, x):
        x = self.pool1(self.relu1(self.bn1(self.conv1(x))))
        x = self.pool2(self.relu2(self.bn2(self.conv2(x))))
        x = self.pool3(self.relu3(self.bn3(self.conv3(x))))
        x = self.flatten(x)
        x = self.dropout(self.relu4(self.fc1(x)))
        x = self.fc2(x)
        return x

model = RegularizedCNN().to(device)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001, weight_decay=1e-4)

# -------------------------------------------------------------
# 4. Kiểm tra nhanh: 1 batch đi qua mô hình
# -------------------------------------------------------------
images, labels = next(iter(train_loader))
print("Kích thước 1 batch ảnh:", tuple(images.shape))
print("Kích thước đầu ra:", tuple(model(images.to(device)).shape))
print("Tổng số tham số:", sum(p.numel() for p in model.parameters()))
```

```python
import time

epochs = 30   # đề dùng 30; máy chậm thì giảm xuống 10

print("\n--- Training Regularized CNN ---")
start = time.time()

for epoch in range(epochs):
    model.train()
    running_loss, correct, total = 0.0, 0, 0

    for images, labels in train_loader:
        images, labels = images.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * images.size(0)
        _, predicted = torch.max(outputs, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

    print(f"Epoch [{epoch+1}/{epochs}] | Loss: {running_loss/len(train_dataset):.4f} "
          f"| Accuracy: {(correct/total)*100:.2f}% | {time.time()-start:.0f}s")

# -------------------------------------------------------------
# Lưu trọng số mô hình
# -------------------------------------------------------------
model_save_path = os.path.join(homework_path, "car_truck_cnn.pth")
torch.save(model.state_dict(), model_save_path)
print(f"\nSuccessfully saved the trained CNN model weights to: {model_save_path}")

# -------------------------------------------------------------
# Đánh giá trên tập validation
# -------------------------------------------------------------
model.eval()
all_preds, all_labels = [], []

with torch.no_grad():
    for images, labels in val_loader:
        images = images.to(device)
        outputs = model(images)
        _, predicted = torch.max(outputs, 1)
        all_preds.extend(predicted.cpu().numpy())
        all_labels.extend(labels.numpy())

print("\n--- Final Classification Report ---")
print(classification_report(np.array(all_labels), np.array(all_preds), target_names=class_names))
```
